## Questão 4 — Diagrama de bifurcação do mapa logístico

$$G(x) = r\,x\,(1 - x)$$

Item (a): construímos o diagrama para $r \in [2.5, 4.0]$ e, em seguida, um *zoom* na janela $r \in [3.73, 3.75]$.

```python
# recebe o x atual e o parâmetro r, devolve o próximo x
def mapa_logistico(x, r):
  x = r*x*(1 - x)
  return x
```

### (a.1) Intervalo $r \in [2.5, 4.0]$

```python
import numpy as np
import matplotlib.pyplot as plt

# intervalo do parâmetro r
rinit = 2.5
rend  = 4.0

# buffers para guardar os pares (r, x) (pontos no gráfico)
buffer_r        = np.zeros(500*100)
buffer_diagrama = np.zeros(500*100)
k = 0

# varre 500 valores de r espaçados igualmente entre rinit e rend
for i in range(0, 500):
  r = rinit + i*(rend - rinit)/500.
  x = 0.1                      # condição inicial

  # iterações *sem guardar*, deixa o sistema "assentar" (descarta o transiente)
  for j in range(0, 500):
    x = mapa_logistico(x, r)

  # aguarda os próximos 100 valores, que mostram o comportamento estável
  for j in range(0 ,100):
    x = mapa_logistico(x, r)
    buffer_r[k] = r
    buffer_diagrama[k] = x
    k = k + 1

# cada ponto é bem pequeno para o padrão de bifurcação ficar nítido
plt.scatter(buffer_r, buffer_diagrama, s=0.01, color='black')
plt.xlabel('r')
plt.ylabel('x')
plt.title('Diagrama de bifurcacao: r em [2.5, 4.0]')
plt.show()
```
